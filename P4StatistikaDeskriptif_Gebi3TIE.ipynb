```python
import pandas as pd
import matplotlib.pyplot as plt

path = "Latihan Penyajian Data Gebi3TIE.xlsx"

dataraw = pd.read_excel(path, header=5, usecols="B:H")
dataraw = dataraw.dropna(how='all')

dataraw.columns = ["ID", "Name", "Quiz", "UTS", "UAS", "Final Score", "Grade"]

dataraw = dataraw.drop(0).reset_index(drop=True)

print("Data Awal (5 baris teratas):")
dataraw.head()
```

```python
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")
print("Tabel Frekuensi Nilai:")
datafrq
```

```python

frekuensi = dataraw["Grade"].value_counts().sort_index()

plt.figure(figsize=(9,5))

plt.plot(
    frekuensi.index,
    frekuensi.values,
    marker='o',
    linewidth=2,
    linestyle='-'
)

plt.title("Distribusi Frekuensi Nilai", fontsize=14)
plt.xlabel("Grade")
plt.ylabel("Jumlah Siswa")

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

frekuensi = dataraw["Grade"].value_counts().sort_index()

plt.figure(figsize=(8,5))
plt.bar(frekuensi.index, frekuensi.values, width=0.5)

plt.title("Jumlah Mahasiswa Berdasarkan Grade")
plt.xlabel("Grade")
plt.ylabel("Jumlah")

plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

```python

frekuensi = dataraw["Grade"].value_counts().sort_index()

plt.figure(figsize=(7,7))

plt.pie(
    frekuensi.values,
    labels=frekuensi.index,
    autopct='%1.0f%%',
    startangle=90
)

plt.title("Persentase Grade Mahasiswa")
plt.show()
```

```python
dataraw["Final Score"] = pd.to_numeric(dataraw["Final Score"], errors='coerce')
dt = dataraw["Final Score"]

stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()

print("Statistika Deskriptif:")
stats
```
